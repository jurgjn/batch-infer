```bash
cd openfold3
# https://github.com/aqlaboratory/openfold-3/blob/main/docker/DOCKER.md#production-images
time docker buildx build --platform=linux/amd64 --progress plain \
    -f docker/Dockerfile \
    --build-arg BUILD_MODE=lock \
    --build-arg CUDA_BASE_IMAGE_TAG=12.1.1-cudnn8-devel-ubuntu22.04 \
    --target devel \
    -t jurgjn/openfold3:$OF3_VERSION .
cd -
```
